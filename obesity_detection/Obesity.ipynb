```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('ObesityDataset.csv')
df.head(10)
```

```python
df.info()
```

```python
df['Obesity Level'] = df['NObeyesdad']
df.drop(columns=['NObeyesdad'], inplace=True)
df
```

```python
df['Age'] = np.round(df['Age'])
df['Height'] = np.round(df['Height'], decimals=2)
df['Weight'] = np.round(df['Weight'])
df
```

```python
df.rename(inplace=True, columns={
    'FAVC': 'eats_high_calorie_food', 
    'FCVC': 'usually_eats_veggies', 
    'NCP': 'daily_main_meals',
    'CAEC': 'food_between_meals',
    'SMOKE': 'smoke',
    'CH2O': 'daily_water',
    'SCC': 'monitor_daily_calories',
    'FAF': 'physical_activity',
    'TUE': 'tech_time',
    'CALC': 'alcohol',
    'MTRANS': 'transport'
})
df
```

```python
df['daily_water'] = np.round(df['daily_water'])
df['physical_activity'] = np.round(df['physical_activity'])
df['tech_time'] = np.round(df['tech_time'])
df
```

```python
sns.countplot(data=df['transport'])
plt.title('Transportation Count')
```

```python
sns.countplot(data=df['Obesity Level'])
plt.title('Obesity Count')
```

```python
sns.countplot(data=df['food_between_meals'])
plt.title('Food Between Meals Count')
```

```python
sns.countplot(data=df['alcohol'])
plt.title('Alcohol Count')
```

```python
sns.countplot(data=df['eats_high_calorie_food'])
plt.title('High Calorie Eaters Count')
```

```python
obesity_mapping = {
    'Insufficient_Weight': 0,
    'Normal_Weight': 1,
    'Overweight_Level_I': 2,
    'Overweight_Level_II': 3,
    'Obesity_Type_I': 4,
    'Obesity_Type_II': 5,
    'Obesity_Type_III': 6
}
df['Obesity Level'] = df['Obesity Level'].map(obesity_mapping)
df
```

```python
history_map = {
    'no': 0,
    'yes': 1
}
df['family_history_with_overweight'] = df['family_history_with_overweight'].map(history_map)

calorie_map = {
    'no': 0,
    'yes': 1
}
df['eats_high_calorie_food'] = df['eats_high_calorie_food'].map(calorie_map)

between_map = {
    'no': 0,
    'Sometimes': 1,
    'Frequently': 2,
    'Always': 3
}
df['food_between_meals'] = df['food_between_meals'].map(between_map)

monitor_map = {
    'no': 1,
    'yes': 0
}
df['monitor_daily_calories'] = df['monitor_daily_calories'].map(monitor_map)

alcohol_map = {
    'no': 0,
    'Sometimes': 1,
    'Frequently': 2,
    'Always': 3
}
df['alcohol'] = df['alcohol'].map(alcohol_map)

smoke_map = {
    'no': 0,
    'yes': 1
}
df['smoke'] = df['smoke'].map(smoke_map)
df
```

```python
gender_mapping = {
    'Female': 0,
    'Male': 1
}
df['Gender'] = df['Gender'].map(gender_mapping)
df
```

```python
df = pd.get_dummies(df, dtype=int)
df
```
